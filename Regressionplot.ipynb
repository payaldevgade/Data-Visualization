```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# REGRESSION is used in machine learning so we will not fully discuss what, how and where
# it is used but we will see how it is created and what are the values we can ut inside it 
```

```python
tips = sns.load_dataset('tips')
```

```python
sns.lmplot(x = 'total_bill', y ='tip', data = tips)
```

```python
sns.lmplot(x = 'total_bill', y ='tip', data = tips,hue = 'sex')
```

```python
sns.lmplot(x = 'total_bill', y ='tip', data = tips, hue = 'sex',palette = 'pastel')
```

```python
sns.lmplot(x = 'total_bill', y ='tip', data = tips, hue = 'sex',palette = 'coolwarm')
```

```python
sns.lmplot(x = 'total_bill', y ='tip', data = tips, hue = 'sex',palette = 'rainbow')
```

```python
sns.lmplot(x='total_bill', y='tip',
           data=tips,
           hue='sex',
           palette='pastel',
           markers=['o', 'v'])

plt.show()
```

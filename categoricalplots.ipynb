```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# CATEGORIAL DATA PLOTS
```

```python
# now let's discuss using seaborn to plot categorical data ! there are a few main plot types for this:
# 1. boxplot
# 2. violinplot
# 3. stripplot
# 4. swarmplot
# 5. barplot
# 6. countplot 
```

```python
df = sns.load_dataset('tips')
df
```

```python
# count plot
```

```python
sns.countplot(df['sex'])
```

```python
sns.countplot(x = df['sex'])
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
# bar plot
```

```python
sns.barplot(x= df['sex'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'])
```

```python
# COMPARISION TYPES
# you can change the estimator object to your own function, that converts a vector to a scalar
# mean (default) : np.mean
# median : np.median
# sum : np.sum
# count : len (to count the number of observations)
# standard deviation : np.std
# minimum : np.min
# maximum : np.max 
```

```python
import numpy as np
sns.barplot(x = df['sex'],y = df['total_bill'],estimator=np.sum)
```

```python
import numpy as np
sns.barplot(x = df['sex'],y = df['tip'],estimator=np.sum)
```

```python
##  box plot
```

```python
sns.boxplot(x = 'tip',y = 'day',data = df)
```

```python
sns.boxplot(x = 'tip',y = 'day',data = df,palette = 'rainbow')
```

```python
sns.boxplot(x = 'tip',y = 'day',data = df,palette = 'pastel')
```

```python
sns.boxplot(x = 'tip',y = 'day',data = df,palette = 'dark')
```

```python
sns.boxplot(x = 'tip',y = 'day',data = df,palette = 'bright')
```

```python
# BOXPLOT AND VIOLINPLOT

# boxplot and violinplot are used to shown the distribution of categorical data 
# A box plot (or box-and-whisker plot) shows the distribution of quantiatative 
 # data in a way that facilitataes comparisons between variables or across levels 
 # of a categorical variables. the box showa the quartiles of the dataset while the 
 # whisker extend to show the rest of the distrinution , except for points that are 
 # determined to be determined to be "outliers" using a method that is a function of the 
 # inter - quartile range
```

```python
# violinplot
```

```python
sns.violinplot(x = df['day'], y ='total_bill',data = df)
```

```python
# STRIPPLOT AND SWARMPLOT

## The stripplot will draw a scatterplot where one variable is categorial . 
# A strip plot can be drawn on its own , but it is a good complement to a box 
# or violin plot in case where you want to show all observations along with some represenattion of the underlying dist

## the swarmplot is similar to stripplot() , but the points are adjusted (only along the catergorial axis) 
# so that they don't overlap . this gives a better representation of the dist of values,
# although it does not scale as well to large no. of observations (both in terms of the ability to show all 
# the points and in terms of the coputation needed to arrange them).
```

```python
# stripplot
```

```python
sns.stripplot(x = 'day', y ='total_bill',data = df)
```

```python
# swarmplot
```

```python
sns.swarmplot(x = 'day', y ='total_bill',data = df)
```

```python
# combine
```

```python
sns.violinplot(x = 'tip',y = 'day', data = df)
```

```python
sns.violinplot(x = 'tip',y = 'day', data = df)
sns.swarmplot(x = 'tip',y = 'day', data = df)
```
